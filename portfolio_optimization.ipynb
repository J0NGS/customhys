### 📌 **Seção 1: Introdução ao Problema de Otimização de Portfólio**
A otimização de portfólio é um problema clássico em finanças, onde buscamos a **melhor distribuição de investimentos** para **maximizar retornos** enquanto **minimizamos riscos**.

#### **Objetivo**
Dado um conjunto de ativos financeiros, queremos encontrar a melhor alocação de recursos para que o portfólio tenha **o maior retorno possível** com **o menor risco possível**.

#### **Como funciona um portfólio de investimentos?**
- Suponha que temos **4 ativos** disponíveis para investimento: 
  1. **Ações da Apple (AAPL)**
  2. **Ações da Google (GOOGL)**
  3. **Ações da Amazon (AMZN)**
  4. **Tesouro Direto (Renda fixa)**

- Para cada ativo, podemos investir um percentual $x_i$, com a **restrição** de que a soma dos investimentos deve ser 100%:

  $$
  x_1 + x_2 + x_3 + x_4 = 1
  $$

  Onde $x_i$ representa a fração do capital investida no ativo $i$.

#### **Medindo Retorno e Risco**
1. **Retorno Esperado ($E[R]$)**: A média dos retornos históricos dos ativos ponderados pelos percentuais $x_i$.
2. **Risco do Portfólio ($\sigma_p$)**: A variabilidade dos retornos (volatilidade), que pode ser modelada como:

   $$
   f(x) = \sum_{i=1}^{n} (\sigma_i x_i)^2
   $$

   Onde $\sigma_i$ é a volatilidade do ativo $i$.

#### **Critério de Otimização**
Queremos **maximizar** o **Sharpe Ratio**, que mede o retorno ajustado ao risco:

$$
SR = \frac{E[R]}{f(x)}
$$

Isso significa que buscamos a **melhor combinação de ativos** para obter o **máximo retorno por unidade de risco**.

---

#### **📊 Exemplo Prático**
Vamos supor que temos **4 ativos financeiros** disponíveis para investimento:
1. **Apple (AAPL)**
2. **Google (GOOGL)**
3. **Amazon (AMZN)**
4. **Tesouro Direto (TD)** (renda fixa)

##### **📈 Retornos Médios dos Ativos**
Os retornos médios dos últimos anos foram os seguintes:

| Ativo  | Retorno Médio Anual ($E[R]$) |
|--------|------------------------------|
| Apple  | 12%  ($0.12$)                |
| Google | 10%  ($0.10$)                |
| Amazon | 14%  ($0.14$)                |
| TD     | 5%   ($0.05$)                |

O **retorno esperado do portfólio** é dado pela fórmula:

$$
E[R_p] = x_1 E[R_1] + x_2 E[R_2] + x_3 E[R_3] + x_4 E[R_4]
$$

Ou seja, é a **média ponderada dos retornos dos ativos** conforme a porcentagem investida em cada um.

**Exemplo de um portfólio inicial:**  
Suponha que começamos com uma alocação arbitrária:
- **30%** em Apple ($x_1 = 0.30$)
- **25%** em Google ($x_2 = 0.25$)
- **35%** em Amazon ($x_3 = 0.35$)
- **10%** em Tesouro Direto ($x_4 = 0.10$)

Agora calculamos o **retorno esperado do portfólio**:

$$
E[R_p] = (0.30 \times 0.12) + (0.25 \times 0.10) + (0.35 \times 0.14) + (0.10 \times 0.05)
$$

$$
E[R_p] = 0.036 + 0.025 + 0.049 + 0.005 = 0.115 \text{ (ou seja, 11,5)}
$$

O retorno médio esperado do portfólio seria **11,5% ao ano**.

---

##### **📉 Medindo o Risco (Função Sphere)**
O **risco do portfólio** pode ser representado pela **volatilidade dos ativos**. Vamos usar o **desvio padrão dos retornos históricos** para medir o risco:

| Ativo  | Risco ($\sigma$) (Desvio Padrão) |
|--------|----------------------------------|
| Apple  | 20%  ($0.20$)                    |
| Google | 18%  ($0.18$)                    |
| Amazon | 25%  ($0.25$)                    |
| TD     | 5%   ($0.05$)                    |

A função **Sphere** é usada para calcular o **risco total** do portfólio:

$$
f(x) = \sum_{i=1}^{n} (\sigma_i x_i)^2
$$

Substituindo os valores:

$$
f(x) = (0.20 \times 0.30)^2 + (0.18 \times 0.25)^2 + (0.25 \times 0.35)^2 + (0.05 \times 0.10)^2
$$

$$
f(x) = (0.06)^2 + (0.045)^2 + (0.0875)^2 + (0.005)^2
$$

$$
f(x) = 0.0036 + 0.002025 + 0.00765625 + 0.000025
$$

$$
f(x) = 0.01330625 \approx 1.33\%
$$

O risco total do portfólio **nesta configuração** é **1.33%**.

---

##### **📊 Sharpe Ratio: Retorno Ajustado ao Risco**
O **índice de Sharpe** nos diz **quanto de retorno obtemos para cada unidade de risco**. Ele é calculado como:

$$
SR = \frac{E[R_p]}{f(x)}
$$

Substituindo:

$$
SR = \frac{0.115}{0.0133} = 8.65
$$

Isso significa que **para cada unidade de risco, obtemos um retorno esperado de 8.65 vezes**.

Nosso objetivo na otimização será encontrar um **portfólio que maximize o Sharpe Ratio**!

#### Criando instâncias que serão utilizadas no exemplo
 Duas instâncias baseadas no exemplo prático que irão ser utilizadas como exemplo no código

```python
import numpy as np
from customhys.benchmark_func import Sphere

# 🎯 Definição da Primeira Instância - 4 Ativos
assets_4 = ["Apple (AAPL)", "Google (GOOGL)", "Amazon (AMZN)", "Tesouro Direto (TD)"]
returns_4 = [0.12, 0.10, 0.14, 0.05]  # Retornos médios anuais
volatilities_4 = [0.20, 0.18, 0.25, 0.05]  # Desvio padrão dos retornos (volatilidade)

portfolio_4_assets = {
    "function": Sphere(4),  # Função Sphere para medir o risco
    "is_constrained": True,  # Restrições de soma dos percentuais = 100%
    "boundaries": ([0] * 4, [1] * 4),  # Cada ativo pode receber de 0% a 100%
    "assets": assets_4,  # Lista de ativos
    "returns": returns_4,  # Retornos médios
    "volatilities": volatilities_4  # Volatilidade dos ativos
}

# 🎯 Definição da Segunda Instância - 10 Ativos
assets_10 = [
    "Apple (AAPL)", "Google (GOOGL)", "Amazon (AMZN)", "Tesla (TSLA)", "Microsoft (MSFT)",
    "Nvidia (NVDA)", "Meta (META)", "Berkshire Hathaway (BRK.B)", "JPMorgan (JPM)", "Tesouro Direto (TD)"
]
returns_10 = [0.12, 0.10, 0.14, 0.15, 0.13, 0.17, 0.11, 0.09, 0.08, 0.05]
volatilities_10 = [0.20, 0.18, 0.25, 0.30, 0.22, 0.27, 0.19, 0.16, 0.14, 0.05]

portfolio_10_assets = {
    "function": Sphere(10),  # Agora para 10 ativos
    "is_constrained": True,
    "boundaries": ([0] * 10, [1] * 10),  # Todos os 10 ativos têm percentuais entre 0% e 100%
    "assets": assets_10,
    "returns": returns_10,
    "volatilities": volatilities_10
}

# ✅ Exibir informações das instâncias criadas
print("Instância 1 - Portfólio com 4 Ativos:", portfolio_4_assets)
print("Instância 2 - Portfólio com 10 Ativos:", portfolio_10_assets)
```

#### Criando função de avaliação
Ela precisa calcular:

- Retorno esperado do portfólio
    - Média ponderada dos retornos históricos dos ativos conforme os percentuais alocados.
- Risco do portfólio
    - Medido pela função Sphere, que soma os quadrados da volatilidade ponderada.
- Índice de Sharpe
    - Maximizar Sharpe Ratio = Retorno esperado / Risco.

No código, exibimos o resultado como negativo, pois por padrão o CUSTOMHYS tenta minimizar o problema, já que nosso objetivo é maximizar nossa função, exibimos o seu resultado como negativo.

```python
from customhys.benchmark_func import Sphere

def portfolio_evaluation(weights, problem, risk_free_rate=0.03):
    print("\n🔍 [LOG] Chamando `portfolio_evaluation` com pesos:", weights)

    returns = np.array(problem["returns"])
    volatilities = np.array(problem["volatilities"])

    #print("[LOG] Retornos:", returns)
    #print("[LOG] Volatilidades:", volatilities)

    weights = np.array(weights)
    weights = weights / np.sum(weights) if np.sum(weights) != 0 else np.full_like(weights, 1 / len(weights))
    print("\n🔍 [LOG] Pesos após normalização:", weights)


    expected_return = np.dot(weights, returns)
    risk = Sphere(len(weights)).get_func_val(weights)

    print("[LOG] Retorno Esperado:", expected_return)
    print("[LOG] Risco Calculado:", risk)

    sharpe_ratio = (expected_return - risk_free_rate) / risk if risk > 0 else -1e6
    resultado = float(-sharpe_ratio)

    print("[LOG] Índice de Sharpe (negativo para minimizar):", resultado)

    return resultado  # Retorna um escalar negativo para ser minimizado pelo CustomHyS

```

##### Testando no portifólio de 4 ativos

```python
# 🔹 Teste com pesos iniciais arbitrários
weights_4 = [0.30, 0.25, 0.35, 0.10]  # Alocação inicial

# 🔹 Cálculo do Índice de Sharpe usando a nova função
sharpe_score = portfolio_evaluation(weights_4, portfolio_4_assets)

print(f"Índice de Sharpe para o portfólio inicial: {-sharpe_score:.2f}")  # Convertendo de negativo para positivo
```

#### Função para criar problema no formato que a hyperheuristica aceita, utilizando os probelmas já definidos acima

```python
def configure_problem(portfolio_instance):
    """
    Gera um dicionário de configuração para a Hyper-Heurística com base em uma instância de portfólio.

    :param portfolio_instance: Dicionário contendo 'function', 'boundaries', 'returns', e 'volatilities'.
    :return: Dicionário formatado corretamente para a Hyper-Heurística.
    """
    return {
        "function": lambda weights: portfolio_evaluation(weights, portfolio_instance),  # Função de avaliação genérica
        "is_constrained": True,  # Garante que a soma dos percentuais respeite 100%
        "boundaries": portfolio_instance["boundaries"],  # Usa os limites definidos na instância
    }
```

```python
# 🔹 Configuração do problema para 4 ativos
problem_4 = configure_problem(portfolio_4_assets)

# 🔹 Configuração do problema para 10 ativos
problem_10 = configure_problem(portfolio_10_assets)

print("Problema 4 ativos:", problem_4)
print("Problema 10 ativos:", problem_10)
```

### 📌 **Seção 3: Encontrando o valor ótimo da instância para comparação**
Nessa seção iremos encontrar o valor ótimo por meio de força bruta, para compararmos com o valor dado pela hyperheuristica

```python
import numpy as np
from scipy.optimize import minimize
from customhys.benchmark_func import Sphere

def find_optimal_portfolio(portfolio_instance, risk_free_rate=0.03):
    """
    Encontra a solução ótima para o problema do portfólio, usando a mesma instância utilizada pela Hyper-Heurística.

    Parâmetros:
    - portfolio_instance (dict): Dicionário contendo "function", "boundaries", "returns" e "volatilities".
    - risk_free_rate (float): Taxa livre de risco para cálculo do Índice de Sharpe.

    Retorna:
    - dict contendo 'optimal_sharpe' e 'optimal_weights'.
    """

    def portfolio_evaluation(weights):
        """
        Função de avaliação do portfólio. Calcula o Índice de Sharpe negativo para minimização.
        """
        weights = np.array(weights)
        weights = weights / np.sum(weights) if np.sum(weights) != 0 else np.full_like(weights, 1 / len(weights))

        returns = np.array(portfolio_instance["returns"])
        expected_return = np.dot(weights, returns)

        risk_function = portfolio_instance["function"]
        risk = risk_function.get_func_val(weights)  # Calculando risco com Sphere

        sharpe_ratio = (expected_return - risk_free_rate) / risk if risk > 0 else -1e6
        return float(-sharpe_ratio)  # Negativo para minimizar

    # Restrição para garantir que a soma dos pesos seja 1
    def constraint_sum(weights):
        return np.sum(weights) - 1

    # Número de ativos na instância do problema
    n_assets = len(portfolio_instance["returns"])

    # Limites definidos na instância do problema
    bounds = list(zip(*portfolio_instance["boundaries"]))

    # Inicialização aleatória dos pesos normalizados
    initial_weights = np.random.rand(n_assets)
    initial_weights /= np.sum(initial_weights)

    # Otimizando usando SLSQP
    result = minimize(
        fun=portfolio_evaluation,
        x0=initial_weights,
        method='SLSQP',
        constraints={'type': 'eq', 'fun': constraint_sum},
        bounds=bounds
    )

    # Retornando o resultado como dicionário
    return {
        "optimal_sharpe": -result.fun,  # Pegamos o valor sem o sinal invertido
        "optimal_weights": result.x
    }
```

```python
# 🔹 Exemplo de uso com o problema configurado
resultado = find_optimal_portfolio(portfolio_4_assets)
print(resultado)
```
